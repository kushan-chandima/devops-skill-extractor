# Analysis of Job Skills for DevOps Engineers

This notebook is used to analyze the extracted job skills data from job postings on LinkedIn and Glassdoor, focusing on both Sri Lanka and international listings.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visualisation style
sns.set(style='whitegrid')
```

## Load the Data

Load the processed job skills data for analysis.

```python
# Load the processed data
data_path = '../data/processed/job_skills_data.csv'
job_skills_df = pd.read_csv(data_path)

job_skills_df.head()
```

## Data Overview

Get an overview of the dataset including the number of entries and the columns present.

```python
job_skills_df.info()
```

## Skills Frequency Analysis

Analyze the frequency of each skill mentioned in the job postings.

```python
skills_frequency = job_skills_df['skills'].value_counts().reset_index()
skills_frequency.columns = ['Skill', 'Frequency']

plt.figure(figsize=(12, 8))
sns.barplot(x='Frequency', y='Skill', data=skills_frequency.head(20))
plt.title('Top 20 Job Skills for DevOps Engineers')
plt.xlabel('Frequency')
plt.ylabel('Skill')
plt.show()
```

## Conclusion

This notebook provides insights into the skills required for DevOps engineers based on job postings. Further analysis can be conducted to explore trends over time or differences between local and international job markets.
